### Using Parameters in Python
How Parameters Work

In a URL, parameters come after a ? and are formatted as key=value pairs:

<https://api-v3.mbta.com/routes?filter[type]=0&include=stop>
                               ↑
                          Parameters start here

```python
# API
BASE_URL = 'https://api-v3.mbta.com/routes'
# Define your parameters in a dictionary
params = {
    "filter[type]": "0",  # Filter for light rail routes only (type 0)
    "include": "stop"     # Include stop information in the response
}

# Pass the params to requests.get()
response = requests.get(BASE_URL, params=params)
print("Status Code:", response.status_code)
print(response.text[:1000])  # Print first 1000 characters
```

```python
# Working with GET Requests and Parameters
BASE_URL = 'https://api-v3.mbta.com/routes'
PARAMS = {
    "filter[type]": "0,1"  # This filters for light rail and heavy rail routes
}

response = requests.get(url=BASE_URL, params=PARAMS)

print(response.text)
```

# REST API 
What is a REST API?
REST stands for `REpresentational State Transfer. `

**Breaking it down:**

* REpresentational: Data is represented in a format (usually JSON)
* State: The current condition of a resource (like a user account, a product, etc.)
* Transfer: Moving that representation between client and server
In simple terms: REST is a set of architectural principles for building web APIs. When an API follows these principles, we call it a "REST API" or "RESTful API."

* GET: Read a list of all objects or one single object by ID.
* POST: Add a new object by sending JSON data.
* PUT / PATCH: Update an existing object.
* DELETE: Remove an object by its ID

```python
import requests

url = "https://api.restful-api.dev/objects"

response = requests.get(url)# Asks the website: “Please give me your data.”

data = response.json()# Converts the website’s answer into Python data.

print(data)
```

```python
#GET means: Give me data.
# POST means: I am sending new data to you.
# This is the new information we want to send.
new_item = {
    "name": "My phone",
    "data": {
        "color": "black",
        "price": 500
    }
}
# url → where we send the information
# json=new_item → the information we send
response = requests.post(url, json=new_item)

created_item = response.json()

print(created_item)
# The API creates the id automatically.
```

```python
# Now let’s learn PATCH.

# PATCH means: Change part of existing data.

# After POST creates an item, the API returns its id.
#  We use that ID to tell the API which item to change.
# Get the ID from the item created with POST
object_id = created_item["id"]
# Information we want to change
change = {
    "name": "My new phone"
}
# Send the change
response = requests.patch(
    f"{url}/{object_id}",
    json=change
)
updated_item = response.json()
print(updated_item)
```

```python
# DELETE means: Remove an existing item.
# We need the item’s id to tell the API which item to delete.
# Get the ID of the item created with POST
object_id = created_item["id"]

# Delete that item
response = requests.delete(f"{url}/{object_id}")

print(response.json())
```

```python
# Step 1: Create main.py
from fastapi import FastAPI # Imports the FastAPI tool.

app = FastAPI() # Creates your API application and names it app.
# Create an application object.

# Creates a GET endpoint.
# The / means the main or home address:
# A decorator that specifies that GET requests made to the application's main URL
# must be handled by this function
@app.get("/Strat Web")
def hello():
    return {"message": "Hello World"}

# Step3: Start the API :uvicorn main:app --reload
#Each part has a meaning:
#uvicorn → starts the server
#main → looks for main.py
#pp → looks for the variable named app
# --reload → restarts automatically when you change the code
```
