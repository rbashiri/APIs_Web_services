```python
import joblib
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris 
from sklearn.model_selection import train_test_split 
# Train your first model
# Load a dataset
iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)
# Train first model 
model= RandomForestClassifier(random_state = 42)
model.fit(X_train, y_train)

# save with version number 
version = 1 
joblib.dump(model, f' iris_classifier_v{version}.joblib')
print(f"Model version {version} saved!")
```

**When you train a new version, you just increment the number:**

```python
# Later, after getting more data or improving your features
iris = load_iris()
X_train_new, X_test_new, y_train_new, y_test_new = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42)

model_v2 = RandomForestClassifier(n_estimators=200, random_state=42)
model_v2.fit(X_train_new, y_train_new)

# Save as version 2
version = 2
joblib.dump(model_v2, 'f iris_classifier_v{version}.joblib')
print(f"Model version {version} saved!")
```
